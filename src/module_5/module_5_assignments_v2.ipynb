```python
# Ensure dependencies are installed
!pip install -q ollama numpy

import ollama
import re

# =====================================================================
# CORE ENGINE: Year-by-Year Chain of Thought Call
# =====================================================================
def chain_of_thought(prompt: str, current_temp: float) -> str:
    """
    Sends a financial scenario prompt to llama3.2:3b with a specific temperature,
    forcing it to calculate and log investment values year-by-year.
    """
    system_instruction = (
        "You are an expert financial calculation engine that uses explicit Chain-of-Thought (CoT) reasoning.\n"
        "When given an investment scenario, you must calculate and log the values YEARLY.\n"
        "For each year, explicitly show, calculate, and write down:\n"
        "1. Starting Balance\n"
        "2. Gross Return\n"
        "3. Annual Fee\n"
        "4. Taxable Growth\n"
        "5. Tax Owed\n"
        "6. Ending Net Balance\n\n"
        "Do not skip any years. At the very end of your response, output a single line formatted EXACTLY like this:\n"
        "FINAL_VALUE: $X.XX"
    )

    response = ollama.chat(
        model='llama3.2:3b',
        options={"temperature": current_temp},  # Varied dynamically per path
        messages=[
            {'role': 'system', 'content': system_instruction},
            {'role': 'user', 'content': prompt}
        ]
    )
    return response.message.content

# =====================================================================
# VERIFICATION METHODS (The Two Calculation Engines)
# =====================================================================
def python_iterative_calc(principal, rate, fee_rate, tax_rate, years):
    """Method 1: Detailed year-by-year loop mapping sequential ledger states"""
    logs = ["\n[Python Engine 1: Iterative Loop Method Log]"]
    current_balance = principal

    for year in range(1, years + 1):
        start_bal = current_balance
        gross_return = start_bal * rate
        fee = start_bal * fee_rate
        taxable_gain = gross_return - fee
        tax = max(0.0, taxable_gain * tax_rate)
        end_bal = start_bal + gross_return - fee - tax

        logs.append(
            f"Year {year}: Start=${start_bal:,.2f} | Return=${gross_return:,.2f} | "
            f"Fee=${fee:,.2f} | Tax=${tax:,.2f} | End=${end_bal:,.2f}"
        )
        current_balance = end_bal

    return round(current_balance, 2), logs

def python_analytical_calc(principal, rate, fee_rate, tax_rate, years):
    """Method 2: Multiplier formula calculation bypassing sequential loops"""
    net_multiplier = 1 + (rate - fee_rate) * (1 - tax_rate)
    final_amount = principal * (net_multiplier ** years)

    logs = [
        "\n[Python Engine 2: Analytical Compound Formula Log]",
        f"Net Multiplier Constant per Annum = {net_multiplier:.6f}",
        f"Calculated Formulaic Value = ${final_amount:,.2f}"
    ]
    return round(final_amount, 2), logs

# =====================================================================
# ORCHESTRATOR: Self-Consistency Engine with Dynamic Auditing
# =====================================================================
def self_consistency_chain_of_thought(prompt: str, n_paths: int = 3) -> str:
    """
    Generates multiple LLM reasoning paths at intentionally varied temperatures,
    runs dual-method 5-year programmatic verifications, parses the LLM answers,
    and directly compares the LLM outputs against the verified target value.
    """
    print(f"🚀 Initializing Self-Consistency Routine (Timeline: 5 Years)...\n")

    # Explicitly configure different temperatures to force varied reasoning path behavior
    temperatures = [0.2, 0.5, 0.7] if n_paths == 3 else [0.3 + (i * 0.2) for i in range(n_paths)]

    llm_outputs = []
    llm_extracted_values = []

    # 1. Sample independent reasoning paths using varied temperatures
    for i in range(n_paths):
        temp = temperatures[i % len(temperatures)]
        print(f"Sampling LLM Reasoning Path {i+1}/{n_paths} (Temperature: {temp})...")
        path_text = chain_of_thought(prompt, current_temp=temp)
        llm_outputs.append(path_text)

        # Parse out the final numerical value from the LLM text output using Regex
        match = re.search(r"FINAL_VALUE:\s*\$?([\d,]+\.\d{2})", path_text)
        if match:
            # Strip commas and cast to float
            val = float(match.group(1).replace(',', ''))
            llm_extracted_values.append(val)
        else:
            llm_extracted_values.append(None)

    # 2. Hardcoded Ground Truth Configurations (Updated strictly for a 5-year calculation gap)
    p, r, f, t, y = 10000, 0.08, 0.01, 0.20, 5

    # 3. Compute Deterministic Checks
    val_method1, logs_m1 = python_iterative_calc(p, r, f, t, y)
    val_method2, logs_m2 = python_analytical_calc(p, r, f, t, y)

    # Verify the code engine calculations match each other
    methods_match = (val_method1 == val_method2)
    verified_target = val_method1

    # 4. Build final report string with explicit audit outputs
    output_report = []
    output_report.append("==================================================")
    output_report.append("=== SECTION 1: REUSED CHAIN-OF-THOUGHT PATHS ===")
    output_report.append("==================================================")
    for idx, path in enumerate(llm_outputs):
        output_report.append(f"\n--- [Path {idx+1} Text Stream | Temp: {temperatures[idx]}] ---")
        output_report.append(path)

    output_report.append("\n==================================================")
    output_report.append("=== SECTION 2: DUAL-METHOD PROGRAMMATIC CHECKS ===")
    output_report.append("==================================================")
    output_report.extend(logs_m1)
    output_report.extend(logs_m2)

    output_report.append("\n==================================================")
    output_report.append("=== SECTION 3: SYSTEM CONSISTENCY MATRIX =======")
    output_report.append("==================================================")
    output_report.append(f"Method 1 (Iterative State Loop):  ${val_method1:,.2f}")
    output_report.append(f"Method 2 (Analytical Multiplier): ${val_method2:,.2f}")
    output_report.append(f"Verification Match Status:        {'✅ VERIFIED MATCH' if methods_match else '❌ DISCREPANCY'}")

    output_report.append("\n--- LLM Path Cross-Audit Performance ---")
    for idx, extracted_val in enumerate(llm_extracted_values):
        if extracted_val is None:
            status = "❌ FAILED TO PARSE FINAL_VALUE"
        elif round(extracted_val, 2) == round(verified_target, 2):
            status = f"✅ ACCURATE (${extracted_val:,.2f})"
        else:
            status = f"❌ MISCALCULATION (${extracted_val:,.2f} vs Target ${verified_target:,.2f})"

        output_report.append(f"• Path {idx+1} Audit Status: {status}")

    return "\n".join(output_report)

# --- Run the Refactored Auditing System ---
investment_prompt = (
    "Calculate the net investment value over a 5-year timeline. "
    "Initial Balance: $10,000. Annual Return: 8%. "
    "Annual Fee: 1% charged on the starting balance of the year. "
    "Tax rate: 20% applied only to the net growth of that year after the fee is subtracted."
)

report = self_consistency_chain_of_thought(investment_prompt, n_paths=3)
print(report)
```

```python
# Install dependencies if needed
!pip install -q ollama

import ollama
from typing import List, Dict

# =====================================================================
# 1. THE SIMPLEMODEL CLASS
# =====================================================================
class SimpleModel:
    def __init__(self, model_name: str = 'llama3.2:3b'):
        self.model_name = model_name
        self.examples: List[Dict[str, str]] = []

    def train_with_examples(self, examples: List[Dict[str, str]]):
        """
        Simulates few-shot training by ingesting input-output pairs into memory.
        Each example should be a dictionary: {'input': '...', 'output': '...'}
        """
        self.examples.extend(examples)
        print(f"🎒 [SimpleModel] Successfully cached {len(examples)} few-shot examples into context memory.")

    def format_few_shot_prompt(self, target_prompt: str) -> str:
        """
        Constructs a structured few-shot prompt combining cached examples
        with the target inference prompt.
        """
        formatted_prompt = "You are a financial calculation engine. Study the examples below and follow their exact structural formatting and calculation logic.\n\n"

        # Append the training examples to build the context
        for i, ex in enumerate(self.examples):
            formatted_prompt += f"### Example {i+1}\n"
            formatted_prompt += f"Input:\n{ex['input']}\n"
            formatted_prompt += f"Output:\n{ex['output']}\n"
            formatted_prompt += "---------------------------------------\n"

        # Append the new target request
        formatted_prompt += "### Target Problem\n"
        formatted_prompt += f"Input:\n{target_prompt}\n"
        formatted_prompt += "Output:\n"

        return formatted_prompt

    def predict(self, full_prompt: str) -> str:
        """Executes inference on the constructed few-shot payload."""
        response = ollama.chat(
            model=self.model_name,
            options={"temperature": 0.0}, # Zero temperature forces strict adherence to example style
            messages=[
                {'role': 'user', 'content': full_prompt}
            ]
        )
        return response.message.content


# =====================================================================
# 2. THE FEW-SHOT ORCHESTRATION FUNCTION
# =====================================================================
def few_shot_prompting(model: SimpleModel, examples: List[Dict[str, str]], prompt: str) -> str:
    """
    Orchestrates the entire few-shot pipeline: ingests examples,
    builds the structured context, and extracts the target prediction.
    """
    # Step 1: "Train" the model with the static templates
    model.train_with_examples(examples)

    # Step 2: Construct the complete prompt sequence
    full_few_shot_context = model.format_few_shot_prompt(prompt)

    # Step 3: Run prediction
    print("🔮 Generating prediction based on few-shot formatting patterns...")
    prediction = model.predict(full_few_shot_context)

    return prediction


# =====================================================================
# 3. RUNNING A FINANCIAL EXECUTION TEST
# =====================================================================

# Define high-quality input-output templates to guide Llama 3.2:3b
financial_training_examples = [
    {
        "input": "Principal: $5,000, Return: 10%, Fee: 2%, Tax: 10%, Years: 1",
        "output": (
            "Year 1:\n"
            "- Start Balance: $5,000.00\n"
            "- Gross Return: $500.00\n"
            "- Fee: $100.00\n"
            "- Tax: $40.00\n"
            "- End Net Balance: $5,360.00\n"
            "FINAL_VALUE: $5,360.00"
        )
    }
]

# The target math sequence we want the model to solve using the identical format
target_scenario_prompt = "Principal: $10,000, Return: 8%, Fee: 1%, Tax: 20%, Years: 5"

# Initialize our simulator
my_model = SimpleModel(model_name='llama3.2:3b')

# Run pipeline
result_output = few_shot_prompting(my_model, financial_training_examples, target_scenario_prompt)
print("\n=== FINAL GENERATED RESPONSE ===")
print(result_output)
```


Core Findings & Technical Insights

This technical implementation demonstrated the construction of a robust, cloud-based financial calculation engine using the Llama 3.2:3b model via Google Colab and Ollama. Operating a lightweight 3-billion-parameter model under strict mathematical scenarios yields the following insights:
• Mitigation of LLM Hallucinations: Smaller language models inherently struggle with multi-step arithmetic operations. By deploying structured Chain of Thought (CoT) system prompts, the model was forced to articulate intermediate variables (Gross Return, Fees, Taxable Gains), drastically reducing cognitive blind spots.
• Architectural Resilience via Self-Consistency: Implementing a multi-path Self-Consistency Engine allowed the architecture to query the LLM multiple times across minor temperature offsets. This isolated random token errors by leveraging independent generation pathways.
• Dual-Engine Deterministic Verification: The validation framework successfully integrated two Python-based programmatic judges:
	1. An Iterative Loop Engine to map sequential ledger transitions.
	2. An Analytical Formula Engine using compound growth constants.
Verifying the open-ended text paths against these deterministic calculations bridges the gap between structured AI reasoning and flawless numeric execution.
• Context Engineering with Few-Shot Classes: Storing static input-output pairs dynamically inside a custom SimpleModel wrapper proved highly effective. At temperature 0.0, it successfully forced the model to mirror syntax, spacing constraints, and specific balance declarations without updating model weights.
